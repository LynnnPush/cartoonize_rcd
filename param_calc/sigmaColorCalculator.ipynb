```python
import math

Q_FRAC = 15
Q_SCALE = 1 << Q_FRAC

def gaussian(diff, sigma):
    return math.exp(-(diff * diff) / (2.0 * sigma * sigma))

def generate_color_lut_fixed(max_diff=255, sigmaColor=100.0):
    print('#include <ap_fixed.h>')
    print()
    print('typedef ap_ufixed<16,1> weight_t;')
    print()
    print(f'static const weight_t COLOR_LUT_FX[{max_diff+1}] = {{')

    for d in range(max_diff + 1):
        w = gaussian(d, sigmaColor)
        fx = int(round(w * Q_SCALE))

        # clamp just in case
        fx = min(fx, Q_SCALE - 1)

        comma = ',' if d < max_diff else ''
        print(f'    weight_t({fx}){comma}')

    print('};')

if __name__ == "__main__":
    generate_color_lut_fixed()
```
