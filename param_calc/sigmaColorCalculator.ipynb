```python
import math

def gaussian(diff, sigma):
    return math.exp(-(diff * diff) / (2 * sigma * sigma))

def generate_color_lut_cpp(max_diff=255, sigmaColor=75.0):
    print(f"constexpr float COLOR_LUT[{max_diff+1}] = {{")

    for d in range(max_diff+1):
        w = gaussian(d, sigmaColor)
        comma = "," if d < max_diff else ""
        print(f"    {w:.8f}{comma}")

    print("};")

# Example usage
if __name__ == "__main__":
    generate_color_lut_cpp(max_diff=255, sigmaColor=75.0)
```
