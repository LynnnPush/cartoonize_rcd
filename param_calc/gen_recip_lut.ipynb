```python
LUT_SIZE = 256

print('#include <ap_fixed.h>')
print()
print('typedef ap_ufixed<18,1> recip_t;')
print()
print(f'static const recip_t RECIP_LUT[{LUT_SIZE}] = {{')

for i in range(LUT_SIZE):
    w = i + 1.0          # avoid divide by zero + smoother
    r = 1.0 / w
    comma = ',' if i < LUT_SIZE - 1 else ''
    print(f'    recip_t({r:.8f}){comma}')

print('};')
```
