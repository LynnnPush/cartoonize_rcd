```python
Q_FRAC = 16
Q_SCALE = 1 << Q_FRAC
LUT_SIZE = 256

print('#include <ap_fixed.h>')
print()
print('typedef ap_ufixed<18,2> recip_t;')
print()
print(f'static const recip_t RECIP_LUT[{LUT_SIZE}] = {{')

for i in range(LUT_SIZE):
    # avoid division by zero
    w = i + 1
    r = 1.0 / w
    fx = int(round(r * Q_SCALE))
    fx = min(fx, (1 << 18) - 1)

    comma = ',' if i < LUT_SIZE - 1 else ''
    print(f'    recip_t({fx}){comma}')

print('};')
```
