```python
import math

def gaussian(dist, sigma):
    return math.exp(-(dist * dist) / (2 * sigma * sigma))

def generate_spatial_cpp(d=5, sigmaSpace=75.0):
    center = d // 2

    # Compute spatial matrix
    spatial = [[0.0 for _ in range(d)] for _ in range(d)]
    for i in range(d):
        for j in range(d):
            dy = i - center
            dx = j - center
            dist = math.sqrt(dx*dx + dy*dy)
            spatial[i][j] = gaussian(dist, sigmaSpace)

    # Generate C++ constexpr output
    print(f"constexpr float SPATIAL_KERNEL[{d}][{d}] = {{")
    for i in range(d):
        row_vals = ", ".join(f"{spatial[i][j]:.8f}" for j in range(d))
        print(f"    {{ {row_vals} }},")
    print("};")

# Example usage
if __name__ == "__main__":
    generate_spatial_cpp(d=11, sigmaSpace=350.0)
```
