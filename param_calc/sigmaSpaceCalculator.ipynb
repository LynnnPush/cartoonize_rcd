```python
import math

Q_FRAC = 15
Q_SCALE = 1 << Q_FRAC

def gaussian(dist, sigma):
    return math.exp(-(dist * dist) / (2.0 * sigma * sigma))

def generate_spatial_fixed(d=5, sigmaSpace=100.0):
    center = d // 2

    print('#include <ap_fixed.h>')
    print()
    print('typedef ap_ufixed<16,1> weight_t;')
    print()
    print(f'static const weight_t SPATIAL_KERNEL_FX[{d}][{d}] = {{')

    for i in range(d):
        row = []
        for j in range(d):
            dy = i - center
            dx = j - center
            dist = math.sqrt(dx*dx + dy*dy)
            w = gaussian(dist, sigmaSpace)
            fx = int(round(w * Q_SCALE))
            fx = min(fx, Q_SCALE - 1)
            row.append(f'weight_t({fx})')

        comma = ',' if i < d - 1 else ''
        print('    { ' + ', '.join(row) + ' }' + comma)

    print('};')

if __name__ == "__main__":
    generate_spatial_fixed()
```
